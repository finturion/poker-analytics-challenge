Waarom dit de moeite waard is: een bot die bijna alles foldt verliest nooit veel, maar wint ook nooit iets. Zonder te tellen weet je niet aan welke kant jouw bot zit.

```python
aantal_raise = 0
aantal_call = 0
aantal_fold = 0

for hand in alle_handen:
    actie = kies_actie(hand)
    if actie == "raise":
        aantal_raise += 1
    elif actie == "call":
        aantal_call += 1
    else:
        aantal_fold += 1

print("raise:", aantal_raise, "->", round(aantal_raise / len(alle_handen) * 100, 1), "%")
print("call: ", aantal_call,  "->", round(aantal_call  / len(alle_handen) * 100, 1), "%")
print("fold: ", aantal_fold,  "->", round(aantal_fold  / len(alle_handen) * 100, 1), "%")
```
