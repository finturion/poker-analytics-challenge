### Het toernooi ophalen


`toernooi_resultaat` is een dictionary met deze velden:

| Veld | Betekenis |
|---|---|
| `n_bots` | hoeveel bots er in dit toernooi meespeelden (inclusief eventuele oefenbots) |
| `namen_deelnemers` | de bot-namen van deze week (meestal gewoon de student_id's) |
| `aangevuld_met_oefenbots` | hoeveel automatische oefenbots zijn toegevoegd omdat er nog te weinig echte inzendingen waren |
| `hand_log` | de volledige, hand-voor-hand log van elke bot: actie, stack, tafel en simulatie |
| `eindstand_per_bot` | de gemiddelde eindstand per bot, samengevat over alle tafels en simulaties |

Verderop in dit onderdeel gebruik je vooral `hand_log` (voor de losse data) en `eindstand_per_bot` (voor het overzicht).


```python
import json

# Dit toernooi is echt gespeeld, door een eerdere lichting. De namen zijn
# verzonnen, de getallen niet: dezelfde stacks, hetzelfde foldgedrag, dezelfde
# uitschieters. Vanaf week 3 haal je je eigen toernooi op bij de API; tot die
# tijd werk je hiermee, zodat je je op het programmeren kunt richten.
with open("data/voorbeeldtoernooi.json") as f:
    toernooi_resultaat = json.load(f)

# In deze uitslag staat jouw bot nog niet -- vanaf week 3 wel. Kies er zolang
# eentje om te volgen; alles hieronder rekent met deze naam. Ravi staat net onder
# het midden, en dat is een eerlijker startpunt dan de nummer 1: de vraag onderaan
# is of je boven of onder de helft zit.
mijn_bot = "Ravi"

print("deelnemers:", len(toernooi_resultaat["namen_deelnemers"]))
print("logregels: ", len(toernooi_resultaat["hand_log"]))
print("je volgt:  ", mijn_bot)
```

Geen internet nodig: het bestand staat naast je notebook in de map `data/`. Vanaf week 3 haal je hier je eigen toernooi op bij de API, en dan zie je je eigen bot in deze uitslag staan.


`toernooi_resultaat` is een gewone dictionary — geen tabel. Je haalt er dus dingen uit met vierkante haken, net als bij de API-antwoorden hierboven:

- `toernooi_resultaat["eindstand_per_bot"]` — per bot zijn eindstand
- `toernooi_resultaat["hand_log"]` — een lijst met één regel per bot per hand

Meer heb je vandaag niet nodig. Volgende week krijg je in Werkcollege 3 pandas erbij, en dan zie je hoe kort dit kan.

**Stap 1 — De stand.** `sorted()` zet een lijst op volgorde. Geef je er een dictionary aan, dan sorteert hij de sleutels; met `key=eindstand.get` sorteert hij ze op hun wáárde, en `reverse=True` zet de hoogste bovenaan.

```python
eindstand = toernooi_resultaat["eindstand_per_bot"]

op_volgorde = sorted(eindstand, key=eindstand.get, reverse=True)

print("de top 5:")
for naam in op_volgorde[:5]:
    print("   ", naam, eindstand[naam])

print()
print(mijn_bot, "staat op:", eindstand[mijn_bot])
print("plek:", op_volgorde.index(mijn_bot) + 1, "van de", len(op_volgorde))
```

🤔 Sta je boven of onder het midden? En belangrijker: zegt één toernooi genoeg om dat te weten?

**Stap 2 — Wat deed jouw bot?** Dit is dezelfde telling als in Werkcollege 1, maar nu over de échte handen in plaats van je 10.000 gesimuleerde. Let op de extra `if`: het `hand_log` bevat álle bots, dus je moet eerst de jouwe eruit pikken.

```python
aantal_raise = 0
aantal_call = 0
aantal_fold = 0

for rij in toernooi_resultaat["hand_log"]:
    if rij["bot_naam"] == mijn_bot:
        if rij["actie"] == "raise":
            aantal_raise += 1
        elif rij["actie"] == "call":
            aantal_call += 1
        else:
            aantal_fold += 1

totaal = aantal_raise + aantal_call + aantal_fold
print("handen gespeeld:", totaal)
print("raise:", aantal_raise, "call:", aantal_call, "fold:", aantal_fold)
```

🤔 Vergelijk met wat je maandag over je 10.000 simulatiehanden telde. Komt de verdeling overeen?

Waarschijnlijk niet helemaal, en dat is logisch: aan een echte tafel krijgt je bot niet elke hand de kans om te doen wat hij wil. Soms is folden de enige geldige zet.
