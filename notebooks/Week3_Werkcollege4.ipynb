### 1. Functies: parameters, return-waarde en default-argumenten (1 min)

Functies met parameters en een return-waarde ken je al uit Week 1 (`kies_actie(hand)`). Nieuw vandaag: **default-argumentwaarden** — een parameter met een standaardwaarde, zodat je 'm niet elke keer hoeft mee te geven. Dat heb je zo meteen in Deel 3 nodig.

**Stap 1 — Gegeven voorbeeld (1 min).** `bereken_pot_odds(pot, inzet)` ken je al. Hieronder een tweede functie die met de uitkomst daarvan iets doet, en meteen een default-argument laat zien: `drempel=30` betekent dat je 'm niet hoeft mee te geven, tenzij je 'm bewust wil overschrijven.

````python
def bereken_pot_odds(pot, inzet):
    return inzet / (pot + inzet) * 100

def is_call_de_moeite_waard(vereiste_winkans, drempel=30):
    """vereiste_winkans: percentage uit bereken_pot_odds(). drempel: max percentage dat je nog accepteert (standaard 30)."""
    return vereiste_winkans <= drempel

winkans = bereken_pot_odds(100, 20)
print(winkans, "->", is_call_de_moeite_waard(winkans))              # gebruikt de default drempel (30)
print(winkans, "->", is_call_de_moeite_waard(winkans, drempel=10))  # een strengere (tight) drempel
````

We bespreken in de klas hoe Bot v2 verschilt van Bot v1: hij krijgt zijn eigen stack mee, en kan straks ook zien in welke ronde van de hand hij zit. Daarmee kan hij voor het eerst op de SITUATIE reageren in plaats van alleen op zijn twee kaarten.

Maar eerst: hoe werkt een hand poker eigenlijk écht?

### De regels van Texas Hold'em (2 min)

- Elke speler krijgt 2 **hole cards**: privé, alleen jij ziet ze. *Hole* betekent hier "in het gat" — ze liggen dicht op tafel, met de rug naar boven. Je hoort ook "pocket cards"; dat is hetzelfde. De Engelse termen staan hier omdat de engine ze gebruikt en omdat alles wat je erover opzoekt ze gebruikt.
- Vóór er ook maar één kaart gedeeld is, leggen 2 spelers verplicht in: de **small blind** en **big blind** (2x zo groot als de small blind). Er zit dus altijd al geld in de pot voordat iemand een keuze heeft gemaakt.
- Er volgen 4 rondes (straten): **preflop** (alleen je hole cards), **flop** (3 gedeelde kaarten erbij), **turn** (1 erbij, 4 totaal), **river** (1 erbij, 5 totaal). Na elke ronde waarin nieuwe kaarten bijkomen, wordt er weer geboden.
- De 5 **community cards** liggen open op tafel, voor iedereen zichtbaar. Aan het einde (**showdown**) maakt elke overgebleven speler de best mogelijke 5-kaarts hand uit zijn 2 hole cards + de 5 community cards samen. Hoogste hand wint de pot.

A heeft **three of a kind**: twee azen in de hand plus de ruiten aas op tafel maakt drie azen. Dat is de kern van Texas Hold'em — je combineert je eigen kaarten *met* het bord, en de beste vijf tellen.

B heeft één paar (de zeven in de hand plus de klaveren zeven op tafel). Maar B heeft ook iets anders: **drie schoppen**. Nog twee erbij en het is een flush.

````python
turn = flop + ["S9"]           # schoppen 9 erbij
river = turn + ["ST"]          # schoppen 10 erbij ("T" van Ten)

for naam, bord in [("TURN ", turn), ("RIVER", river)]:
    print(naam, bord)
    print("  A:", beschrijf_hand(speler_a, bord).ljust(18),
          "B:", beschrijf_hand(speler_b, bord).ljust(18),
          "-> wint:", vergelijk_handen(speler_a, speler_b, bord))
````

### Welke acties kan jouw bot kiezen? (1 min)

Aan het begin van elke ronde wordt jouw `kies_actie`-functie aangeroepen. Je geeft één van deze strings terug:

| actie | wat het doet | kosten |
|---|---|---|
| `"fold"` | je legt je hand weg | 0, maar je verliest wat er al in de pot ligt van jou |
| `"call"` | je legt bij tot het bedrag dat op tafel ligt | het verschil; 0 als niemand nog geboden heeft (een gratis "check") |
| `"raise"` | je verhoogt met het kleinste toegestane bedrag | zie hieronder |
| `"grote_raise"` | je verhoogt naar 200 chips | tien big blinds |
| `"all_in"` | je zet je hele resterende stack in | alles |

`"grote_raise"` en `"all_in"` mag je pas gebruiken vanaf deze week, en alleen als `stack` in je functie staat — logisch, want zonder te weten hoeveel je hebt kun je niet afwegen hoeveel je inzet.

### Hoeveel is een raise? (2 min)

## Deel 6 — Pot odds × winkans: een echte beslisregel (12 min)

Nu heb je beide helften van dezelfde vraag:

- `bereken_pot_odds(pot, inzet)` → hoeveel procent je **minimaal moet** winnen om een call rendabel te maken.
- `schat_winkans(hand)` → hoeveel procent je **naar verwachting wint**.

> ⚠️ **`schat_winkans` kijkt alleen naar je twee eigen kaarten.** Hij kent het bord niet, en dat kán ook niet: op het moment dat je bot beslist, ligt de flop er nog niet. Wat hij doet is duizend keer een willekeurig bord uitdelen en tellen hoe vaak je wint. Dat is dus een **preflop-schatting** — een gemiddelde over alle borden die nóg kunnen komen.
>
> Concreet: `A K` heeft ongeveer 66% en dat blijft 66%, ook als de flop drie tweeën brengt en je hand in werkelijkheid kansloos is. Vanaf de flop weet je meer dan die tabel. Dat is geen fout in de functie, het is de grens ervan — en de reden dat je in je bot na preflop iets anders nodig hebt.

En daarmee heb je de klassieke pokerregel te pakken:

> **Call als je winkans hoger is dan wat de pot odds van je eisen. Anders fold.**

Dat is geen vuistregel maar rekenwerk: als de pot 100 is en je moet 20 bijleggen, heb je 20/(100+20) ≈ 17% nodig. Met `A K` (≈66% winkans) is callen dan duidelijk goed; met `7 2` (≈34%) nog steeds, en dát is precies waarom "altijd folden met slechte kaarten" ook niet klopt.

**Stap 1 — Gegeven voorbeeld.**

````python
# bereken_pot_odds heb je in Deel 1 al geschreven -- die hergebruiken we hier
def is_call_verstandig(hand, pot, inzet_om_te_callen):
    if inzet_om_te_callen == 0:
        return True  # gratis meedoen, altijd prima
    vereiste_winkans = bereken_pot_odds(pot, inzet_om_te_callen)
    return schat_winkans(hand, seed=1) > vereiste_winkans

for hand, pot, inzet in [(["A", "K"], 100, 20), (["7", "2"], 100, 20), (["7", "2"], 100, 300)]:
    nodig = round(bereken_pot_odds(pot, inzet), 1)
    print(hand, f"| pot {pot}, bijleggen {inzet} -> nodig {nodig}%,",
          f"jij hebt {schat_winkans(hand, seed=1)}% ->", "CALL" if is_call_verstandig(hand, pot, inzet) else "FOLD")
````

### Vijf richtingen om te verkennen

Je hoeft ze niet alle vijf te doen. Kies wat bij jouw idee van een goede bot past, en werk dat dan ook echt uit.

1. **Andere drempels per straat.** Preflop weet je alleen je twee kaarten; op de river ligt alles op tafel. Een bot die overal dezelfde grens gebruikt, gooit die informatie weg.
2. **Je winkans-grens bepalen.** Vanaf welke winkans call je, en vanaf welke raise je? Dat is een getal dat je kunt kiezen — en waarvan je kunt méten of het beter werkt dan een ander getal.
3. **De pot-odds-regel, en wanneer je ervan afwijkt.** De regel uit Deel 6 is wiskundig correct maar naïef: hij foldt bijna nooit preflop bij een kleine inzet. Wil je dat?
4. **Stack-bewustzijn.** Met 90 chips speel je anders dan met 900. En je mag nu bewust `"all_in"` teruggeven — wanneer is dat een goed idee in plaats van een noodgreep?
5. **Je twee raises per straat.** Je mag per straat maximaal twee keer verhogen. Waar zet je die in: meteen druk zetten, of pas als je weet wat het bord doet? En als iemand jou verhoogt — pak je terug of laat je het lopen?
6. **Hoeveel je inzet.** Je hebt nu twee maten: `"raise"` (het minimum, meestal 20 erbij) en `"grote_raise"` (200 in één keer). Dat is een echte keuze. Groot inzetten duwt mensen van middelmatige handen af, maar je verliest ook meer als je verkeerd zit. Wanneer is dat het waard?

    (["K", "K"], 1000, "river", 800, 0),        # sterk, gratis meedoen
]
