Dictionaries: https://datalab01.ict.hva.nl/w/pDj3qijdsQeNZPr8cXsk65

Maak een dictionary `bot_profiel` met de keys `naam`, `strategie`, `stack` en `aantal_handen_gespeeld`.

Schrijf daarna een functie `print_profiel(profiel)` die deze nette, leesbare regels print (niet de rauwe dictionary-repr).


```python
bot_profiel = {
    "naam": "vul_hier_je_student_id_in",
    "strategie": "tight",
    "stack": 1000,
    "aantal_handen_gespeeld": 0,
}

def print_profiel(profiel):
    for key, waarde in profiel.items():
        print(f"{key}: {waarde}")

print_profiel(bot_profiel)
```
